```python
def chai_bill(cups, *addons, **discount):
    bill = cups * 15
    bill += len(addons) * 5

    if "discount_percent" in discount:
        bill -= bill * (discount["discount_percent"] / 100)

    return bill


print(chai_bill(2))
print(chai_bill(3, "ginger", "masla"))
print(chai_bill(4, "biscuit", discount_percent=10))
print(chai_bill(5, "ginger", "masala", "biscuit", discount_percent=10))
```

```python
class YoutubeChannel:

    def __init__(self, name):
        self.name = name
        self.subscribers = 0

    def subscribe(self):
        self.subscribers += 1
        print(f"Thanks for subscribing! Total: {self.subscribers}")

    def unsubscribe(self):
        if self.subscribers > 0 :
            self.subscribers -= 1
            print(f"Total: {self.subscribers}")

    def show(self):
        print(f"{self.name} has {self.subscribers} subscribers")


codebasics = YoutubeChannel("codebasics")
codebasics.subscribe()
codebasics.subscribe()       
codebasics.subscribe()
codebasics.unsubscribe()
codebasics.show() 

AIcode = YoutubeChannel("AIcode")
AIcode.subscribe()
AIcode.show()
```

```python
class ShoppingCart:

    def __init__(self,owner):
        self.owner = owner
        self.items = []

    def add_item(self, item):
        self.items.append(item)
        print(f"Added {item}. Cart has {len(self.items)} item(s).")

    def remove_item(self, item):
        if item in self.items:
            self.items.remove(item)
            print(f"Removed {item}. Cart has {len(self.items)} item(s).")
        else:
            print(f"{item} not found in cart.")

    def show_cart(self):
        print(f"{self.owner}'s cart: {self.items}")

cart = ShoppingCart("Farzeen")
cart.add_item("Notebook")
cart.add_item("Pen")
cart.remove_item("Pen")
cart.remove_item("Laptop")
cart.show_cart()
```

```python
def safe_divide(a,b):
    try:
        a = float(a)
        b = float(b)
        result = a/b
        return result

    except ValueError:
        return "Please enter valid numbers."

    except ZeroDivisionError:
        return "Cannot divide by zero."

print(safe_divide(10, 2)) 
print(safe_divide(10, 0)) 
print(safe_divide("ten", 2))
print(safe_divide("20", "4"))
```

### Exercise: Movie-Recommendation Buddy

```python
# Step 1: Define everything (conversation list, chat function, reset function)

conversation = []

def chat(movie_recommendation):
    conversation.append({"role": "user", "parts": [{"text":movie_recommendation}]})
    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents=conversation
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"

    conversation.append({"role": "model", "parts": [{"text": reply}]})
    return reply

def reset():
    global conversation
    conversation = []
    print("Conversation reset! Starting fresh.")

# Step 2: Have your movie conversation
print(chat("I love sci-fi and thrillers, but I can't stand horror movies."))
print(chat("Suggest me a movie to watch tonight."))
print(chat("Why did you pick that one?"))

# Step 3: Done with that topic? Call reset() to start fresh
reset()

# Step 4: Now start a completely NEW, unrelated conversation
print(chat("What's a good recipe for beginners?"))
```

```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

from pydantic import BaseModel
from google.genai import types

class Book(BaseModel):
    title:str
    pages:int

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Suggest me some urdu novel books of Nemrah Ahmad",
    config=types.GenerateContentConfig(
        response_mime_type="application/json",    
        response_schema=Book,
    ),
                 
)
book = response.parsed
print(book.title)
print(book.pages)
```

```python
class Task(BaseModel):
    name: str
    priority: int
    done: bool

tasks = ["Finish report", "Clean inbox", "Call client"]

for task in tasks:
    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=f"Analyze this task: {task}",
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=Task,
        ),
    )
    result = response.parsed
    print(f"{result.name} - Priority {result.priority} Done: {result.done}")
```

```python
class WeatherReport(BaseModel):
    city: str
    temperature: int
    condition: str

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Give me today's weather for Lahore",
    config=types.GenerateContentConfig(
        response_mime_type="application/json",
        response_schema=WeatherReport,
    ),
)
weather = response.parsed
print(weather.city)
```

```python
class Product(BaseModel):
    name: str
    price: float
    in_stock: bool

products_data = ["laptop", "phone", "headphones"]

results = []
for item in products_data:
    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=f"Give me product details for: {item}",
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=Product,
        ),
    )
    results.append(response.parsed)

for product in results:
    print(product.name, product.price)
```

### Exercise: Build a growing recipe collection using JSON

```python
import os
if os.path.exists("recipes.json"):
    os.remove("recipes.json")
print("Cleaned up — starting fresh.")
```

```python
import json
import os

def save_recipe(recipe):
    filename = "recipes.json"

    if os.path.exists(filename):
        with open(filename, "r", encoding="utf-8") as f:
            recipes = json.load(f)
    else:
        recipes = []

    titles = [r["title"] for r in recipes]

    if recipe["title"] not in titles:
        recipes.append(recipe)
        with open(filename, "w", encoding="utf-8") as f:
            json.dump(recipes, f, indent=2)  
        print(f"Saved! You now have {len(recipes)} recipe(s) in your collection.")
    else:
        print(f"{recipe['title']} is already saved — skipping.")

recipe1 = {
    "title": "Yellow Dal",
    "calories": 280,
    "ingredients": ["lentils", "onion", "tomato", "cumin"]
}

recipe2 = {
    "title": "Chicken Karahi",
    "calories": 450,
    "ingredients": ["chicken", "tomato", "green chili", "ginger garlic paste"]
}
recipe3 = {
    "title": "Mixed Rice",
    "calories": 350,
    "ingredients": ["rice", "potato", "peas", "carrot", "onion", "cumin seeds", "garam masala"]
}

save_recipe(recipe1)   
save_recipe(recipe2)
save_recipe(recipe3)

with open("recipes.json", "r", encoding="utf-8") as f:
    all_recipes = json.load(f)

for r in all_recipes:
    print(f"- {r['title']} ({r['calories']} cal)")
```
