```python
class YoutubeChannel:

    def __init__(self, name):
        self.name = name
        self.subscribers = 0

    def subscribe(self):
        self.subscribers += 1
        print(f"Thanks for subscribing! Total: {self.subscribers}")

    def unsubscribe(self):
        if self.subscribers > 0 :
            self.subscribers -= 1
            print(f"Total: {self.subscribers}")

    def show(self):
        print(f"{self.name} has {self.subscribers} subscribers")


codebasics = YoutubeChannel("codebasics")
codebasics.subscribe()
codebasics.subscribe()       
codebasics.subscribe()
codebasics.unsubscribe()
codebasics.show() 

AIcode = YoutubeChannel("AIcode")
AIcode.subscribe()
AIcode.show()
```

```python
class ShoppingCart:

    def __init__(self,owner):
        self.owner = owner
        self.items = []

    def add_item(self, item):
        self.items.append(item)
        print(f"Added {item}. Cart has {len(self.items)} item(s).")

    def remove_item(self, item):
        if item in self.items:
            self.items.remove(item)
            print(f"Removed {item}. Cart has {len(self.items)} item(s).")
        else:
            print(f"{item} not found in cart.")

    def show_cart(self):
        print(f"{self.owner}'s cart: {self.items}")

cart = ShoppingCart("Farzeen")
cart.add_item("Notebook")
cart.add_item("Pen")
cart.remove_item("Pen")
cart.remove_item("Laptop")
cart.show_cart()
```

```python
def safe_divide(a,b):
    try:
        a = float(a)
        b = float(b)
        result = a/b
        return result

    except ValueError:
        return "Please enter valid numbers."

    except ZeroDivisionError:
        return "Cannot divide by zero."

print(safe_divide(10, 2)) 
print(safe_divide(10, 0)) 
print(safe_divide("ten", 2))
print(safe_divide("20", "4"))
```

### Exercise: Movie-Recommendation Buddy

```python
# Step 1: Define everything (conversation list, chat function, reset function)

conversation = []

def chat(movie_recommendation):
    conversation.append({"role": "user", "parts": [{"text":movie_recommendation}]})
    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents=conversation
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"

    conversation.append({"role": "model", "parts": [{"text": reply}]})
    return reply

def reset():
    global conversation
    conversation = []
    print("Conversation reset! Starting fresh.")

# Step 2: Have your movie conversation
print(chat("I love sci-fi and thrillers, but I can't stand horror movies."))
print(chat("Suggest me a movie to watch tonight."))
print(chat("Why did you pick that one?"))

# Step 3: Done with that topic? Call reset() to start fresh
reset()

# Step 4: Now start a completely NEW, unrelated conversation
print(chat("What's a good recipe for beginners?"))
```

### Exercise: Build a growing recipe collection using JSON

```python
import os
if os.path.exists("recipes.json"):
    os.remove("recipes.json")
print("Cleaned up — starting fresh.")
```

```python
import json
import os

def save_recipe(recipe):
    filename = "recipes.json"

    if os.path.exists(filename):
        with open(filename, "r", encoding="utf-8") as f:
            recipes = json.load(f)
    else:
        recipes = []

    titles = [r["title"] for r in recipes]

    if recipe["title"] not in titles:
        recipes.append(recipe)
        with open(filename, "w", encoding="utf-8") as f:
            json.dump(recipes, f, indent=2)  
        print(f"Saved! You now have {len(recipes)} recipe(s) in your collection.")
    else:
        print(f"{recipe['title']} is already saved — skipping.")

recipe1 = {
    "title": "Yellow Dal",
    "calories": 280,
    "ingredients": ["lentils", "onion", "tomato", "cumin"]
}

recipe2 = {
    "title": "Chicken Karahi",
    "calories": 450,
    "ingredients": ["chicken", "tomato", "green chili", "ginger garlic paste"]
}
recipe3 = {
    "title": "Mixed Rice",
    "calories": 350,
    "ingredients": ["rice", "potato", "peas", "carrot", "onion", "cumin seeds", "garam masala"]
}

save_recipe(recipe1)   
save_recipe(recipe2)
save_recipe(recipe3)

with open("recipes.json", "r", encoding="utf-8") as f:
    all_recipes = json.load(f)

for r in all_recipes:
    print(f"- {r['title']} ({r['calories']} cal)")
```

```python
import requests
def get_exchange_rate(base, target):
    try:
        url = f"https://api.frankfurter.dev/v2/rate/{base}/{target}"
        response = requests.get(url)
        print(response.status_code)
        data = response.json()
        print(data)
        return data["rate"]
    except Exception as e:
        print(f"Something went wrong: {e}")
        return None
```

```python
def convert_and_log(base, target):
    rate = get_exchange_rate(base, target)
    if rate is None:
        print("Couldn't load the rate. Log in failed.")
        return

    record = {
        "base": base,
        "target": target,
        "rate": rate,
        "date": str(datetime.now())
    }

    if os.path.exists("conversion_history.json"):
        with open("conversion_history.json", "r", encoding="utf-8") as file:
            history = json.load(file)
    else:
        history = []

    history.append(record)

    with open("conversion_history.json", "w", encoding="utf-8") as file:
        json.dump(history, file, indent=2)

    print(f"Logged: {base} -> {target} at rate {rate}")

convert_and_log("EUR", "PKR")
convert_and_log("USD", "GBP")
```

### Task: Personal Expense Tracker (CLI)

```python
import os
import json
from pydantic import BaseModel, ValidationError

class Expense(BaseModel):
    amount: float
    category: str
    date: str

def add_expense(amount, category, date):
    try:
        expense = Expense(amount=amount, category=category, date=date)
        return expense
    except ValidationError:
        print("Invalid expense")
        return None

def save_expenses(expense):
    expense_dict = expense.model_dump()
    
    if os.path.exists("expenses.json"):
        with open("expenses.json", "r", encoding="utf-8") as file:
            expenses = json.load(file)
    else:
        expenses = []
    
    expenses.append(expense_dict)
    
    with open("expenses.json", "w", encoding="utf-8") as file:
        json.dump(expenses, file, indent=2)
    
    print("Expense saved.")


e1 = add_expense(12.50, "food", "2026-09-10")
save_expenses(e1)

e2 = add_expense(30.00, "transport", "2026-09-10")
save_expenses(e2)

```

### Topic: Database Interaction (SQLite)

```python
import sqlite3
 
# ============================================================
# 1. CONNECT ONCE - reuse this conn/cursor everywhere below
# ============================================================
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
 
# ============================================================
# 2. CREATE TABLE (safe to re-run thanks to IF NOT EXISTS)
# ============================================================
cursor.execute("""
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        grade INTEGER
    )
""")
conn.commit()
 
# ============================================================
# 3. CLEAN SLATE - wipe old data so we start fresh, no duplicates
# ============================================================
cursor.execute("DELETE FROM students")
conn.commit()
 
# ============================================================
# 4. INSERT - multiple rows at once with executemany()
# ============================================================
students = [
    ("Ali", 85),
    ("Sara", 95),
    ("Taha", 75),
    ("Zara", 72),
    ("Imran", 60)
]
cursor.executemany(
    "INSERT INTO students (name, grade) VALUES (?, ?)",
    students
)
conn.commit()
 
# ============================================================
# 5. READ - SELECT all rows
# ============================================================
cursor.execute("SELECT * FROM students")
print("All students:", cursor.fetchall())
 
# READ with WHERE - find one student by name
cursor.execute("SELECT * FROM students WHERE name = ?", ("Imran",))
print("Imran's record:", cursor.fetchall())
 
# ============================================================
# 6. UPDATE - change Taha's grade
# ============================================================
cursor.execute("UPDATE students SET grade = ? WHERE name = ?", (75, "Taha"))
conn.commit()
 
cursor.execute("SELECT * FROM students WHERE name = ?", ("Taha",))
print("Taha after update:", cursor.fetchall())
 
# ============================================================
# 7. DELETE - remove one student by name
# ============================================================
cursor.execute("DELETE FROM students WHERE name = ?", ("Ali",))
conn.commit()
 
cursor.execute("SELECT * FROM students")
print("After deleting Ali:", cursor.fetchall())
 
# ============================================================
# 8. SORTING - ORDER BY
# ============================================================
cursor.execute("SELECT * FROM students ORDER BY grade")
print("Sorted ascending:", cursor.fetchall())
 
cursor.execute("SELECT * FROM students ORDER BY grade DESC")
print("Sorted descending:", cursor.fetchall())
 
# ============================================================
# 9. COUNTING - COUNT(*) with fetchone() (only 1 row possible)
# ============================================================
cursor.execute("SELECT COUNT(*) FROM students")
result = cursor.fetchone()   # fetchone(), not fetchall() - only one row ever comes back
print("Total students:", result[0])
 
cursor.execute("SELECT COUNT(*) FROM students WHERE grade > ?", (80,))
result = cursor.fetchone()
print("Students with grade > 80:", result[0])
 
# ============================================================
# 10. GROUP BY - separate table to practice aggregation
# ============================================================
cursor.execute("""
    CREATE TABLE IF NOT EXISTS enrollments (
        id INTEGER PRIMARY KEY,
        student_name TEXT NOT NULL,
        subject TEXT NOT NULL,
        score INTEGER
    )
""")
conn.commit()
 
cursor.execute("DELETE FROM enrollments")
conn.commit()
 
enrollments = [
    ("Ali", "Math", 85),
    ("Sara", "Math", 95),
    ("Taha", "Math", 70),
    ("Ali", "Science", 60),
    ("Sara", "Science", 88),
    ("Zara", "Science", 92)
]
cursor.executemany(
    "INSERT INTO enrollments (student_name, subject, score) VALUES (?, ?, ?)",
    enrollments
)
conn.commit()
 
# Average score per subject
cursor.execute("""
    SELECT subject, AVG(score)
    FROM enrollments
    GROUP BY subject
""")
print("Average score per subject:", cursor.fetchall())
 
# Count of students per subject
cursor.execute("""
    SELECT subject, COUNT(*)
    FROM enrollments
    GROUP BY subject
""")
print("Students per subject:", cursor.fetchall())
 
# Highest score per subject
cursor.execute("""
    SELECT subject, MAX(score)
    FROM enrollments
    GROUP BY subject
""")
print("Top score per subject:", cursor.fetchall())

# ============================================================
# 11. Combining two related tables (e.g., your students and enrollments tables) 
# ============================================================
cursor.execute("""
    SELECT students.name, students.grade, enrollments.subject, enrollments.score
    FROM students
    JOIN enrollments ON students.name = enrollments.student_name
        WHERE enrollments.score > ?
""", (85,))
print(cursor.fetchall())
 
# ============================================================
# 12. CLOSE - only when fully done with the database
# ============================================================
conn.close()
 
```
