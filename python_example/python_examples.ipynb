```python
financials = {
    "Q1" : {"revenue" : 50, "expenses": 10}, 
    "Q2" : {"revenue" : 60, "expenses": 20},     
    "Q3" : {"revenue" : 55, "expenses": 30},
    "Q4" : {"revenue" : -45, "expenses": 30}
}
for quarter, data in financials.items():
    if data["revenue"]<0:
        print(f"Invalid revenue for {quarter}. Skipping.")
        continue
    margin = data["revenue"] - data["expenses"] * 100 / data["revenue"]
    print(f"{quarter} : {margin:.2f}%")
  
```

```python
def chai_bill(cups, *addons, **discount):
    bill = cups * 15
    bill += len(addons) * 5

    if "discount_percent" in discount:
        bill -= bill * (discount["discount_percent"] / 100)

    return bill


print(chai_bill(2))
print(chai_bill(3, "ginger", "masla"))
print(chai_bill(4, "biscuit", discount_percent=10))
print(chai_bill(5, "ginger", "masala", "biscuit", discount_percent=10))
```

```python
class YoutubeChannel:

    def __init__(self, name):
        self.name = name
        self.subscribers = 0

    def subscribe(self):
        self.subscribers += 1
        print(f"Thanks for subscribing! Total: {self.subscribers}")

    def unsubscribe(self):
        if self.subscribers > 0 :
            self.subscribers -= 1
            print(f"Total: {self.subscribers}")

    def show(self):
        print(f"{self.name} has {self.subscribers} subscribers")


codebasics = YoutubeChannel("codebasics")
codebasics.subscribe()
codebasics.subscribe()       
codebasics.subscribe()
codebasics.unsubscribe()
codebasics.show() 

AIcode = YoutubeChannel("AIcode")
AIcode.subscribe()
AIcode.show()
```

```python
class ShoppingCart:

    def __init__(self,owner):
        self.owner = owner
        self.items = []

    def add_item(self, item):
        self.items.append(item)
        print(f"Added {item}. Cart has {len(self.items)} item(s).")

    def remove_item(self, item):
        if item in self.items:
            self.items.remove(item)
            print(f"Removed {item}. Cart has {len(self.items)} item(s).")
        else:
            print(f"{item} not found in cart.")

    def show_cart(self):
        print(f"{self.owner}'s cart: {self.items}")

cart = ShoppingCart("Farzeen")
cart.add_item("Notebook")
cart.add_item("Pen")
cart.remove_item("Pen")
cart.remove_item("Laptop")
cart.show_cart()
```

```python
def safe_divide(a,b):
    try:
        a = float(a)
        b = float(b)
        result = a/b
        return result

    except ValueError:
        return "Please enter valid numbers."

    except ZeroDivisionError:
        return "Cannot divide by zero."

print(safe_divide(10, 2)) 
print(safe_divide(10, 0)) 
print(safe_divide("ten", 2))
print(safe_divide("20", "4"))
```

### Exercise: Movie-Recommendation Buddy

```python
conversation = []

def chat(movie_recommendation):
    conversation.append({"role": "user", "parts": [{"text":movie_recommendation}]})
    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents=conversation
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"

    conversation.append({"role": "model", "parts": [{"text": reply}]})
    return reply

print(chat("I love sci-fi and thrillers, but I can't stand horror movies."))
print(chat("Suggest me a movie to watch tonight."))
print(chat("Why did you pick that one?"))
```

```python
def reset():
    global conversation
    conversation = []
    print("Conversation reset! Starting fresh.")
```

```python
print(chat("My favorite color is blue."))
```

```python
print(chat("What's my favorite color?"))  
```

```python
reset()
print(chat("What's my favorite color?"))
```
