```python
def average(numbers):  
    # Functions with list parameters (using sum(), len())
    total = sum(numbers)
    count = len(numbers)
    return total / count

result = average([10,20,30,40,50,60])
print(result)
```

```python
def power(base, exponent=2):
    return base ** exponent
print(power(5))
print(power(5,3))
```

```python
def average(num):
    total = sum(num)
    count = len(num)
    return total / count
output = average([10,20,30])
print(output)
```

```python
# Returning Multiple Values
def min_max(integers):
    smallest = min(integers)
    biggest = max(integers) 
    return smallest, biggest

low, high = min_max([10, 25, 3, 47, 8])
print(low)
print(high)
```

```python
# *args — variable number of positional arguments
def add_all(*args):
    print(args)          # args is a tuple of whatever was passed in
    return sum(args)

print(add_all(1, 2, 3))        # 6
print(add_all(5, 10, 15, 20))  # 50
```

```python
# **kwargs — variable number of keyword (named) arguments

def show_info(**kwargs):
    print(kwargs)          # kwargs is a dictionary

show_info(name="Ali", age=21, city="Lahore")
```

```python
def total_price(*args):
    print(args)
    return sum(args)

total_price(100, 250, 75)
```

```python
def describe_person(**kwargs):
    print(kwargs)

describe_person(name="Farzeen", age=20, city="Sheikhupura")
```

```python
def calculate_margin(revenue, expense):
    profit = revenue - expense
    margin = profit * 100 / revenue
    return margin
calculate_margin(80, 10)
```

```python
def calculate_profit_margin(revenue, expense):
    profit = revenue - expense
    margin = profit * 100 / revenue
    return profit, margin

revenue = 80
expense = 10
profit, margin = calculate_profit_margin(revenue, expense)
```

```python
def chai_bill(cups, *addons, **discount):
    bill = cups * 15
    bill += len(addons) * 5

    if "discount_percent" in discount:
        bill -= bill * (discount["discount_percent"] / 100)

    return bill


print(chai_bill(2))
print(chai_bill(3, "ginger", "masla"))
print(chai_bill(4, "biscuit", discount_percent=10))
print(chai_bill(5, "ginger", "masala", "biscuit", discount_percent=10))
```
