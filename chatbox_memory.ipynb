```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()
```

```python
conversation = []   # this list IS the chatbot's memory

def chat(user_message):
    conversation.append({"role": "user", "parts": [{"text": user_message}]})

     # 2. send the ENTIRE conversation so far
    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents=conversation
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"

        # 3. add the AI's reply to memory too, so it remembers next time
    conversation.append({"role": "model", "parts": [{"text": reply}]})
    return reply

print(chat("My name is Farzeen."))
```

### Bonus 1: reset() function

```python
conversation = []

def chat(user_message):
    conversation.append({"role": "user", "parts": [{"text": user_message}]})
    
    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents=conversation
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"

    conversation.append({"role": "model", "parts": [{"text": reply}]})
    return reply

def reset():
    global conversation                  # tells Python: modify the REAL outer 'conversation' list,
                                         # not create a new local one that disappears after this function ends
    conversation = []                    # empty the list — this IS what "forgetting" means here 
    print("Conversation reset! Starting fresh.")


print(chat("My name is Farzeen."))

reset()                              # Call reset() anytime you want to start a brand new, unrelated conversation
print(chat("Suggest me some cooking recipes!"))        #reset() then chat("...") will have NO memory of anything said before
```

### Bonus 2: Give Your Bot a Personality

```python
from google.genai import types
# 'types' is a MODULE (toolbox) that lives inside the google.genai packag

config = types.GenerateContentConfig(
# GenerateContentConfig is a TOOL inside that toolbox — calling it builds a config object

    system_instruction="You are a witty movie buff who loves making puns."  
# system_instruction shapes HOW the AI behaves/talks, for every message — it works BEHIND THE SCENES, unlike normal chat messages
)

conversation = []
def chat(user_message):
    conversation.append({"role": "user", "parts": [{"text": user_message}]})

# 'config' is just a variable name I chose myself — it now holds that personality setting

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents=conversation,
            config=config      # <- this line applies the personality to this call
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"

    conversation.append({"role": "model", "parts": [{"text": reply}]})
    return reply

print(chat("My name is Farzeen Fatima."))
reset()
print(chat("Suggest me a movie to watch tonight."))

# IMPORTANT: config is SEPARATE from conversation.
# reset() only wipes 'conversation' (memory) — it does NOT reset 'config' (personality).
# That's why the witty/pun tone kept showing up even after calling reset().
```
