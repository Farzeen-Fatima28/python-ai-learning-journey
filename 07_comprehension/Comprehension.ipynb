## List Comprehensions

A shortcut to build a list in **one line**, instead of a full loop with `.append()`.

### Syntax breakdown
```python
[expression for item in iterable]
```
- `expression` → what to do with each item (`x ** 2`)
- `for item in iterable` → the loop you already know
- `[ ]` around it → builds a list automatically

### Adding a filter condition
```python
even_squares = [x ** 2 for x in range(1, 11) if x % 2 == 0]
```

````python
# Old way (loop)
squares = []
for x in range(1, 6):
    squares.append(x ** 2)

# New way (list comprehension) — same result
squares = [x ** 2 for x in range(1, 6)]
squares
````

````python
cube = [ x ** 3 for x in range(1,6)]
cube
````

````python
words = ["cat", "elephant", "dog", "python", "ai"]

long_words = [word for word in words if len(word)>4]
long_words
````

### 1.Lengths of only the long words

````python
[len(word) for word in words if len(word) > 4]
````

### 2: Uppercase the long words

````python
words = ["cat", "elephant", "dog", "python", "ai"]

uppercase_long_words = [word.upper() for word in words if len(word) > 4]
uppercase_long_words
````

### : Long words that also start with a vowel

````python
vowels = "aeiouAEIOU"
vowel_words = [word for word in words if len(word) > 4 and word[0] in vowels]
vowel_words
````

### ## Dictionary Comprehensions

Same idea as list comprehensions, but builds a DICTIONARY {} instead of a list [].
Every item becomes a KEY : VALUE pair, instead of just one value.

### Syntax
```python
{key_expression: value_expression for item in iterable if condition}
```
- `key_expression`   → what becomes the dictionary KEY
- `value_expression` → what becomes the dictionary VALUE for that key
- `for item in iterable` → the normal loop, same as list comprehensions
- `if condition` → optional filter, same as list comprehensions

### Example — square numbers as a dictionary
```python
squares_dict = {x: x**2 for x in range(1, 6)}
# {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
# x is the key, x**2 is the value
```

### Example — even numbers and their squares (with filter)
```python
numbers = [12, 7, 33, 4, 18, 25, 9, 40, 15, 3]
even_squares_dict = {x: x**2 for x in numbers if x % 2 == 0}
# {12: 144, 4: 16, 18: 324, 40: 1600}
# only even numbers pass the filter and become keys
```

### Key difference from list comprehension
- List comprehension →  [ expression        for item in iterable if condition ]
- Dict comprehension  →  { key : value       for item in iterable if condition }

The ONLY real difference is using { } instead of [ ], and writing
"key : value" instead of a single expression.

````python
squares_dict = {x: x**2 for x in range(1, 6)}
squares_dict
````

````python
words = ["cat", "elephant", "dog", "python", "ai"]
word_lengths = {word: len(word) for word in words if len(word) > 4}
word_lengths
````
