```python
from google import genai
client = genai.Client()

# 1. Define the function with its 3 parameters:

def generate_recipe(ingredients, cuisine, diet):

# 2. Add the docstring (the '''...''' explanation right after):
    '''
    Generate a recipe using ingredients provided.
    Cuisine can be "Italian", "Mexican", "Indian", "Pakistani" etc.
    diet can be "Vegan", "Vegeterian", "Jain", "Any" etc.
    '''

#3. Build the prompt using an f-string:    
    prompt = f'''
    Generate one food receipe using these ingredients: {','.join(ingredients)}
    Receipe should not be more than 100 words
    Cuisine: {cuisine}
    Diet: {diet}
    '''

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )
    return response.text
```

```python
print(generate_recipe(["dal", "onion", "tomatoes"], "Indian", "Vegan"))
```

```python
print(generate_recipe(["Corns", "black Beans", "Avocado"], "Mexican", "Vegan"))
```
