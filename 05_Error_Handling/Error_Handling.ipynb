### try / except — Error Handling

```python
try:
    # code that might fail
    result = 10 / 0    
except ZeroDivisionError:
     # runs only if something goes wrong above
    print("You can't divide by zero!")
```

```python


def generate_recipe(ingredients, cuisine, diet):
    '''
    Generate a recipe using ingredients provided.
    Cuisine can be "Italian", "Mexican", "Indian", "Pakistani" etc.
    diet can be "Vegan", "Vegeterian", "Jain", "Any" etc.
    '''
    
    prompt = f'''
    Generate one food receipe using these ingredients: {','.join(ingredients)}
    Receipe should not be more than 100 words
    Cuisine: {cuisine}
    Diet: {diet}
    '''
    try:
        from google import genai
        client = genai.Client()
        response = client.models.generate_content(
            model="gemini-9.9-fake",
            contents=prompt
        )
        return response.text
            
    except Exception as e:
        return f"Sorry, couldn't generate a recipe right now. Error: {e}"

print(generate_recipe(["dal", "onion", "tomatoes"], "Indian", "Vegan"))
```

```python
def safe_chai_bill(cups, price_per_cup):
    try:
        cups = int(cups)
        price_per_cup = float(price_per_cup)

        if cups < 1:
            return "Cups must be at least 1"
        else:
            totalbill = cups * price_per_cup
            return totalbill
    except ValueError as e:
        return f"Please enter number only"

print(safe_chai_bill("3", "15"))
print(safe_chai_bill(2, 15))
print(safe_chai_bill("two", 15))
print(safe_chai_bill(0, 15))
    
```
