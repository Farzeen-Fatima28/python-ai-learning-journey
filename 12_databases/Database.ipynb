## SQLite Basics — Reviews Database

This notebook builds a small SQLite database (`shop.db`) that stores
customer reviews with their sentiment analysis (label + score).

## Steps covered
1. **Connect** to the database file and get a cursor
2. **Create** a `reviews` table (id, text, label, score)
3. **Insert** reviews — one at a time and in bulk (`executemany`)
4. **Read** all reviews back with `SELECT *`
5. **Filter** reviews with `WHERE label = ?`
6. **Clean up** duplicate/typo data with `DELETE` and `DROP TABLE`

```python
import sqlite3
```

```python
conn = sqlite3.connect("shop.db")

cursor = conn.cursor()

print("database is ready")
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS reviews (
         id INTEGER PRIMARY KEY,
         text TEXT,
         label TEXT,
         score INTEGER
    )
""")
```

```python
cursor.execute("DELETE FROM reviews")
conn.commit()
```

```python
more_reviews = [
    ("The biryani was amazing.", "positive", 5),
    ("Delivery was late and food is cold.", "negative", 1),
    ("Good taste, fair price.", "positive", 4),
    ("It was okay nothing special.", "neutral", 3),
    ("App kept crashing during payment.", "negative", 2)
]
cursor.executemany(
    "INSERT INTO reviews (text, label, score) VALUES (?, ?, ?)",
    more_reviews
)
conn.commit()
```

```python
cursor.execute("SELECT * FROM reviews")
#print("All reviews:", cursor.fetchall())
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute("SELECT text, score FROM reviews WHERE label=?", ("negative",))
negatives = cursor.fetchall()

print("Negative reviews:")
for text, score in negatives:
    print("-", text, "(score:", score, ")")
conn.commit()
```

```python
conn.close
print("Database closed")
```

```python
import sqlite3

conn = sqlite3.connect("products.db")

cursor = conn.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS products (
         id INTEGER PRIMARY KEY,
         name TEXT,
         price INTEGER
    )
""")
cursor.execute("DELETE FROM products")
conn.commit()

product_list = [
    ("Tea", 10),
    ("Coffee", 25),
    ("Notebook", 30)
]
        
cursor.executemany(
    "INSERT INTO products (name, price) VALUES (?, ?)",
    product_list
)

cursor.execute("SELECT * FROM products")
print("Product_List:", cursor.fetchall())

cursor.execute("SELECT * FROM products WHERE price > ?", (20,)) 
expensive = cursor.fetchall()

for row in expensive:
    print(row)

conn.commit()
    
```
