```python
revenue = [50, 60, 55]
expenses = [10,20,30]
```

```python
for rev in revenue:
    print(f"revenue: {rev}")
```

```python
for i in range(len(revenue)):
    print(f"rev: {i} : {revenue[i]}")
```

```python
for i in range(len(expenses)):
    print(f"exp: {i} : {expenses[i]}")
```

```python
for i in range(len(revenue)):
    profit = revenue[i] - expenses[i]
    margin = profit * 100 / revenue[i]
    print(f"profit: {i} : {profit}")
    print(f"margin: {i} : {margin:.2f}")
```

```python
margins = []
for i in range(len(revenue)):
    profit = revenue[i] - expenses[i]
    margin = profit * 100/ revenue[i]
    margins.append(margin)

margins
```

```python
profits = []
for i in range(len(revenue)):
    profit = revenue[i] - expenses[i]
    profits.append(profit)
    total_profit = sum(profits)
print(f"Total profit: {total_profit}")

profits
```

### Nested Dictionary

```python
financials = {
    "Q1" : {"revenue" : 50, "expenses": 10}, 
    "Q2" : {"revenue" : 60, "expenses": 20},     
    "Q3" : {"revenue" : 55, "expenses": 30},
}
financials
```

```python
financials["Q1"]["revenue"]
```

```python
for quarter, data in financials.items():
    print("Quarter: ", quarter)
    print("Data: ", data)
    
```

```python
for x in financials:                 #without .items()
    print(x, ":", financials[x])
```
