```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

# put in a prompt and ask
prompt = f'''
Here's a Python refrence guide
{document}

explain it in a simple and in bullet point
'''
try:
    response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt
    )
    summary = response.text
except Exception as e:
    summary = f"something went wrong: {e}"

print(summary)
```

```python
with open("summary.txt", "w", encoding="utf-8") as f:
    f.write(summary)

print("Summary is saved to summary.txt")
```

### Saving a dictionary to a JSON file

```python
import json

recipe = {
    "title": "Yellow Dal",
    "calories": 280,
    "ingredients": ["lentils", "onion", "tomato", "cumin"]
}

# json.dump(recipe, f, indent=2) → converts your Python dictionary into proper JSON text, and writes it directly into the file f
with open("recipe.json", "w", encoding="utf-8") as f:
    json.dump(recipe, f, indent=2)             
    
print("recipe.json saved.")

# json.load(f) → reads the JSON file and converts it back into a real Python dictionary
with open("recipe.json", "r", encoding="utf-8") as f:
    loaded = json.load(f)

loaded
```

# Exercise: Meeting Notes → Action Items

## Exercise: Meeting Notes → Action Items

**Goal:** Build a tool that reads messy meeting notes and turns them
into a saved to-do list.

### Steps:
1. Reuse your `python.txt` file (the one with the production meeting
   transcript from earlier)
2. Read it, and ask the LLM to extract the action items — who needs
   to do what
3. Save the result to `action_items.txt`

### Bonus (ties everything together):
Instead of plain text, ask for **structured output** (a Pydantic model
with a list of `{task, owner}`), then save it with `json.dump()`.
Load it back and print each task. Now you've combined file handling +
structured output into one mini-app.

```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()
```

```python
# Step 1: Read the meeting notes
with open("python.txt", "r", encoding = "utf-8") as f:
    document = f.read()
print(document)

# Step 2: Ask the LLM to extract action items
prompt = f"""
Here is a Python reference guide:
{document}

Summarize this into a short action list — what topics should someone
practice or review next, based on this content.
"""

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt
)

action_items = response.text

# Step 3: Save the result
with open("action_items.txt", "w", encoding="utf-8") as f:
    f.write(action_items)

print("Saved to action_items.txt!")
print(action_items)
```
