```python
class Car:

    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
        
    def start(self):
        print(f"{self.brand} is starting!")


car1 = Car("toyota", "Red")
car2 = Car("Honda", "Blue")
print(car1.brand, car2.color)

car1.start()
car2.start()
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def calculate_bonus(self):
        return self.salary * 0.10

emp1 = Employee("Farzeen", 45000)
emp2 = Employee("Hamna", 50000)
emp1.name
emp2.salary
emp1.calculate_bonus()
```

```python
class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

    def display(self):
        print(f"{self.name} (Roll no: {self.roll_no})")

std1 = Student("Farzeen Fatima", 110810)
std1.name
std1.display()
```

```python
# Looping through multiple objects

std1 = Student("Farzeen Fatima", 110810)
std2 = Student("Ali Raza", 110822)
std3 = Student("Sara Khan", 110835)

students = ([std1, std2, std3])

for s in students:
    s.display()
```

## Modules & Imports — Key Takeaways

- A **module** = a Python file full of reusable code (functions, classes, variables)
- Python comes with many **built-in modules** ready to use

### Three ways to import:

- `import module` → access everything with `module.something`
  (e.g. `import math` → `math.sqrt(16)`)

- `from module import thing` → import just what you need, no prefix
  (e.g. `from math import sqrt` → `sqrt(16)`)

- `import module as alias` → give it a shorter nickname
  (e.g. `import math as m` → `m.sqrt(16)`)

### Useful built-in modules:
- `math` → `sqrt`, `pi`, `floor`, `ceil`
- `random` → `randint(a, b)`, `choice(list)`
- `datetime` → `datetime.now()`, `.strftime()`, date differences with `.days`

### Common trap:
- `randint` is a **function inside** `random`, not its own module
  - ❌ `import randint`
  - ✅ `from random import randint` or `import random` → `random.randint(...)`

```python
import random
number = random.randint(1, 100)
number
```

```python
from random import randint as r
number = r(1, 100)
number
```

```python
from datetime import datetime
class Human:
    def __init__(self, name, birth_year):
      self.name = name
      self.birth_year = birth_year
        
    def get_age(self):
      current_year = datetime.today().year
      return current_year - self.birth_year

    def introduce(self):
      print(f"Hello, I'm {self.name} & I'm {self.get_age()} years old!")

person1 = Human("Rabia Noor", 2004)
person2 = Human("Kinza Gul", 2004)
person3 = Human("Farzeen Fatima", 2006)

persons = ([person1, person2, person3])
for p in persons:
    p.introduce()

```

## Pydantic Key Takeaway

- **Normal class** → trusts you, stores whatever you give it (even wrong types)
- **Pydantic model** → checks you, enforces type hints as real rules
- Converts reasonable values (`"280"` → `280`)
- Rejects invalid ones (`"two hundred"` → error)

```python
from pydantic import BaseModel

class Recipe(BaseModel):
    title: str                  # must be a string
    calaroies: int              # must be a number (int) - or convertible to one

dal1 = Recipe(title ="Yellow Dal" , calaroies= 280)
dal1.title                      #'Yellow Dal'

# ✅ This ALSO works - Pydantic auto-converts "280" (string) into 280 (int)
# because it's a valid number written as text (this is called "type coercion")

dal2 = Recipe(title="Palak Paneer", calaroies="280")

# ❌ This FAILS immediately with a ValidationError
# because "two hundred" can't be converted into a real integer
# --> dal3 = Recipe(title="Yellow Dal", calories="two hundred")
```
